```python
list_1 = [1, 2, 3, 4, 5]
list_2 = [2, 3, 4, 5, 6]
list_3 = [9, 7, 6, 8, 9] 
array_3 = np.array([list_1,list_2,list_3])
array_3
```

```python
array_3.shape
```

```python
# In numpy [a,b] - a represents rows and b represents columns. 
# So, if we want to select 1st row and all columns then we will write [0,:] because 0 represents 1st row and : represents all columns. 
# If we want to select 2nd row and 1st column then we will write [1:2,0:1] because 1:2 represents 2nd row and 0:1 represents 1st column. 
# If we want to select 2nd and 3rd row and 1st column then we will write [1:3,0:1] because 1:3 represents 2nd and 3rd row and 0:1 represents 1st column.
```

```python
array_3[0,:]
```

```python
array_3[1:2,0:1]
```

```python
array_3[1:3,0:1]
```

```python
array_3[:,:]
```

```python
array_3[0:2,0:2]
```

Explanation: here we are selecting 1st and 2nd row and 1st and 2nd column. 
So, we will write [0:2,0:2] because 0:2 represents 1st and 2nd row and 0:2 represents 1st and 2nd column.


```python
array_3[1:2,2:3]
```

```python
array_3[1,2]
```
